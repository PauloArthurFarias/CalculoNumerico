### Tema 2:
Seja um movimento físico regido pela função f(d) = a*ed – 4*d2

, onde a são amplitudes dadas devido à
oscilação encontrada em cada movimento considerado e d é o deslocamento encontrado em cada movimento
considerado, variando com o valor de a. Caso esse deslocamento passe de 0,7 cm esse movimento amplifica,
causando sérios danos e um problema gigantesco. O método de Newton modificado é tal que a função de
iteração j(x) é dada por j(x) = x – (f(x) / f ’(xo)), onde xo é uma aproximação inicial e é tal que f ’(xo) 1 0.
Desenvolva um sistema para calcular o valor do deslocamento d que deve atender a todos requisitos abaixo:
a) Implementar algoritmo para calcular d pelo método da Newton-Raphson.
b) Implementar algoritmo para calcular d pelo método de Newton modificado.
c) Implementar algoritmo para calcular d pelo método da Secante original.
d) Testar os resultados para d usando como padrão a = 1, d0 = 0,5 e e = 10-4
.
e) Fornecer um quadro resposta, com d calculado para cada método dado.
f) Fornecer um quadro comparativo, com todos os dados para cada método.
g) Analisar o efeito da variação do valor de a para cada método considerado.
Dados de entrada: n (número de valores de a), a (para cada n) e e (precisão).
Dados de saída: quadros resposta (com d e erro para cada a e método) e comparativo.

```python
import numpy as np
```

### Definindo função do problema

```python
CONSTANTE_A = 1
CONSTANTE_D0 = 0.5
CONSTANTE_E = 0.0001

def f_d(d, a=CONSTANTE_A):
    return a * np.exp(d) - 4 * pow(d, 2)
```

### Definindo e testando derivada numérica

```python
def derivada_numérica(function, x):
    h = 0.0001
    return (function((x + h)) - function(x)) / h
```

```python
def quadratica(x):
    return pow(x, 2) - 9

def poli(x):
    return 2 * pow(x, 3) + 5 * pow(x, 2) + 3

def ln(x):
    return np.log(x)

derivada_numérica(quadratica, 3), derivada_numérica(poli, 10), derivada_numérica(ln, 2)
```

### Definindo e testando Newton-Raphson

```python
def newton_raphson(function, d=CONSTANTE_D0, a = CONSTANTE_A, erro = CONSTANTE_E, iter = 0):
    if abs(function(d)) < erro: return (d, f'Número de iterações: {iter}')
    if iter >= 40: return ('Raíz não encontrada', iter)
    return newton_raphson(function, d = d - (function(d)/derivada_numérica(function, d)),  iter=iter+1)
```

```python
print(newton_raphson(f_d))
```

### Definindo e testando Secante

```python
def metodo_secante(function, d1, d=CONSTANTE_D0, a = CONSTANTE_A, erro = CONSTANTE_E, iter=0):
    if abs(function(d1)) < erro: return (d, f'Número de iterações: {iter}')
    if iter >= 40: return ('Raíz não encontrada')
    d1_temporario_auxiliar = d1
    return metodo_secante(function, d1 = (d * function(d1) - d1 * function(d)) / (function(d1) - function(d)), d = d1_temporario_auxiliar, iter=iter+1)
```

```python
print(metodo_secante(f_d, 1))
```

### Definindo e testando Newton Modificado

```python
derivada_fx0 = derivada_numérica(f_d, CONSTANTE_D0)

def newton_modificado(function, d=CONSTANTE_D0, a = CONSTANTE_A, erro = CONSTANTE_E, derivada = derivada_fx0, iter = 0):
    if abs(function(d)) < erro: return (d, f'Número de iterações: {iter}')
    if iter >= 40: return ('Raíz não encontrada')
    return newton_modificado(function, d = d - (function(d) / derivada), iter=iter+1)
```

```python
newton_modificado(f_d)
```
